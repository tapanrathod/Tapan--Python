#### 14. Create an array from 1 to 30. Reshape it into all possible valid 2D shapes and print them.


```python
# 14. Create an array from 1 to 30. Reshape it into all possible valid 2D shapes and print them.

import numpy as np

original_array = np.arange(1, 31)

total_elements = original_array.size

print(f"Original Array: {original_array}\n")
print("All possible valid 2D shapes and their corresponding reshaped arrays:")

for i in range(1, total_elements + 1):
    if total_elements % i == 0:  # Check if i is a factor of total_elements
        rows = i
        cols = total_elements // i
        reshaped_array = original_array.reshape(rows, cols)
        print(f"Shape ({rows}, {cols}):\n{reshaped_array}\n")
```

#### 24. Write a program that repeatedly asks the user for reshape dimensions until a valid reshape is entered.


```python
# 24. Write a program that repeatedly asks the user for reshape dimensions until a valid reshape is entered.

import numpy as np

total_elements = int(input("Enter the total number of elements in the array : "))
original_array = np.arange(1, total_elements + 1)

while True:
    try:
        rows = int(input("Enter the number of rows for reshape : "))
        cols = int(input("Enter the number of columns for reshape : "))
        reshaped_array = original_array.reshape(rows, cols)

        print(f"\nReshaped Array ({rows}x{cols}):\n{reshaped_array}")
        print(f"Shape of Reshaped Array: {reshaped_array.shape}")
        
        break  # Exit the loop if reshape is successful

    except ValueError:
        print(f"\nError: Cannot reshape array of size {total_elements} into shape ({rows}, {cols}). Please check your input and try again.")
```

#### 27. Create a menu-driven program:

#### 1. Show Shape

#### 2. Show Dimensions

#### 3. Reshape Array

#### 4. Exit


```python
# 27. Create a menu-driven program:
# 1. Show Shape
# 2. Show Dimensions
# 3. Reshape Array
# 4. Exit

import numpy as np

def show_shape(array):
    print(f"Shape of the array: {array.shape}")

def show_dimensions(array):
    print(f"Number of dimensions (ndim): {array.ndim}")

def reshape_array(array):

    try:
        rows = int(input("Enter the number of rows for reshape : "))
        cols = int(input("Enter the number of columns for reshape : "))
        reshaped_array = array.reshape(rows, cols)

        print(f"\nReshaped Array ({rows}x{cols}):\n{reshaped_array}")

    except ValueError:
        print(f"\nError: Cannot reshape array of size {array.size} into shape ({rows}, {cols}). Please check your input and try again.")

def main():

    original_array = np.arange(1, 17)  # Example array from 1 to 16

    while True:
        print("\nMenu:")
        print("1. Show Shape")
        print("2. Show Dimensions")
        print("3. Reshape Array")
        print("4. Exit")

        choice = input("Enter your choice (1-4): ")

        if choice == '1':
            show_shape(original_array)
        elif choice == '2':
            show_dimensions(original_array)
        elif choice == '3':
            reshape_array(original_array)
        elif choice == '4':
            print("Exiting the program.")
            
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 4.")

if __name__ == "__main__":
    main()
```

#### 28. Take dimensions from the user and calculate whether:

#### rows × columns = total elements

#### If yes, reshape the array; otherwise display an error.


```python
# 28. Take dimensions from the user and calculate whether:
# rows × columns = total elements
# If yes, reshape the array; otherwise display an error.

import numpy as np

total_elements = int(input("Enter the total number of elements in the array : "))
original_array = np.arange(1, total_elements + 1)

rows = int(input("Enter the number of rows for reshape : "))
cols = int(input("Enter the number of columns for reshape : "))

if rows * cols == total_elements:
    reshaped_array = original_array.reshape(rows, cols)

    print(f"\nReshaped Array ({rows}x{cols}):\n{reshaped_array}")

else:
    print(f"\nError: Cannot reshape array of size {total_elements} into shape ({rows}, {cols}). Please check your input.")
```

#### 30.Create a complete reshape validator:

#### - Take array size from the user.

#### - Create array using np.arange().

#### - Take any number of dimensions from the user.

#### - Check if multiplication of dimensions equals total elements.

#### - If valid, reshape and print shape and ndim.

#### - Otherwise display a meaningful error message.


```python
# 30.Create a complete reshape validator:
# - Take array size from the user.
# - Create array using np.arange().
# - Take any number of dimensions from the user.
# - Check if multiplication of dimensions equals total elements.
# - If valid, reshape and print shape and ndim.
# - Otherwise display a meaningful error message.

import numpy as np

array_size = int(input("Enter the size of the array : "))
original_array = np.arange(1, array_size + 1)

try:
    num_dimensions = int(input("Enter the number of dimensions for reshape : "))
    dimensions = []

    for i in range(num_dimensions):
        dim = int(input(f"Enter size for dimension {i + 1} : "))
        dimensions.append(dim)

    if np.prod(dimensions) == array_size:
        reshaped_array = original_array.reshape(dimensions)

        print(f"\nReshaped Array with dimensions {tuple(dimensions)}:\n{reshaped_array}")
        print(f"Shape: {reshaped_array.shape}")
        print(f"Number of dimensions (ndim): {reshaped_array.ndim}")

    else:
        print(f"\nError: The product of the specified dimensions {dimensions} does not equal the total number of elements ({array_size}). Please check your input.")

except ValueError:
    print("Invalid input. Please enter valid integers for array size and dimensions.")
```
