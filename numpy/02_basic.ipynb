### Array Creation by given range

```python
arr = np.arange(2, 11, 2)
arr
```

```python
arr = np.arange(11)
arr
```

```python
# change the shape :

arr2 = np.arange(1, 11)
print(arr2.shape)
arr3 = arr2.reshape(5, 2, 1)
arr3
```

```python
arr2.reshape(5, 1, 2)
```

```python
arr4 = np.arange(1, 17)
arr4
```

```python
print(arr4.reshape(4, 2, 2))
print(arr4.reshape(4, 2, 2).shape)
print(arr4.reshape(4, 2, 2).ndim)
```
