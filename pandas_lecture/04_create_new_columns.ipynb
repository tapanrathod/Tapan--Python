```python
import pandas as pd
```

```python
df = pd.read_csv("./dataset/employee_dataset.csv")
df.head()
```

```python
df["salary_bonus"] = (df["salary"] * 10) / 100
df.head()
```

```python
df["city_deparment"] = df["city"] + "-" + df["department"]
df.head()
```

```python
def salary_category(x):

    if x >= 100000:
        return "High"
    
    elif x >= 50000:
        return "Medium"
    
    elif x >= 20000:
        return "Low"


df["salary_category"] = df["salary"].apply(salary_category)
df
```

```python
df["salary"].min(), df["salary"].max()
```

```python
df.groupby("salary_category")["salary_category"].count()
```

```python
df.groupby(["salary_category", "city"])["salary_category"].count()
```

```python
df.to_csv("upgraded_employee_dataset.csv")
```
