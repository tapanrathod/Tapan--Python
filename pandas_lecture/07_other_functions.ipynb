#### Pandas Datetime

```python
df = pd.read_csv("./dataset/sales_data.csv") # reads csv
df
```

```python
df.dtypes # data types of columns in the table
```

```python
df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y") # it wont change the date formate because its in the standard ISO 8601 format
df
```

```python
df["month"] = df["Date"].dt.month # to show only month column
df
```

```python
df["new_date_format"] = df["Date"].dt.strftime("%d-%m-%Y") # to change the date formate
df
```

#### Query

```python
df.query("Quantity >= 3") # query must be written in string aka "" (double quotes)
```

```python
n = 3 # "n" is variable here

df.query("Quantity >= @n") # @ for variables
```

```python
df.query("Product == 'Mobile'") # for a specific product
# here product name must be in '' (singe quotes)
```

```python
p = ["Mobile", "Keyboard"] # "p" is variable here

df.query("Product in @p")
```

#### Crosstab

```python
pd.crosstab(index=df["Gender"], columns=df["Product"], values=df["Price"], aggfunc="sum", margins=True, margins_name="Total")

# index # column name # values #aggfunc must to operate the data

# here margins for sum and margins name for rename the new column and row

# same as pivot table
```
