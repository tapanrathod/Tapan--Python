#### 1. Find the total number of missing values in each column.

```python
# 1. Find the total number of missing values in each column.

import pandas as pd

df = pd.read_csv("../../dataset/payments.csv")

df.isnull().sum()
```

#### 2. Fill missing 'amount' values using city-wise median.

```python
# 2. Fill missing 'amount' values using city-wise median.

df['amount'] = df['amount'].fillna(df.groupby('city')['amount'].transform('median'))

df
```

#### 3. Remove rows where 'name' or 'gender' is missing


```python
# 3. Remove rows where 'name' or 'gender' is missing.

df.dropna(subset=['name', 'gender'], inplace=True)

df
```

#### 4. Create a pivot table showing total amount spent per month by gender

```python
# 4. Create a pivot table showing total amount spent per month by gender.

df.pivot_table(values='amount', index='month', columns='gender', aggfunc='sum')

df
```

#### 5. Create a pivot table showing average spending per category in each city

```python
# 5. Create a pivot table showing average spending per category in each city.

df.pivot_table(values='amount', index='category', columns='city', aggfunc='mean')

df
```

#### 6. Create a pivot table to show count of transactions per payment method per month

```python
# 6. Create a pivot table to show count of transactions per payment method per month.

df.pivot_table(values='amount', index='payment_method', columns='month', aggfunc='count')

df
```

#### 7. Create a multi-level pivot table with index = (city, gender) and columns = month, showing total amount

```python
# 7. Create a multi-level pivot table with index = (city, gender) and columns = month, showing total amount.

df.pivot_table(values='amount', index=['city', 'gender'], columns='month', aggfunc='sum')

df
```

#### 8. Find the top 5 customers based on total spending

```python
# 8. Find the top 5 customers based on total spending.

df.groupby('name')['amount'].sum().nlargest(5)

df
```

#### 9. Create a pivot table showing total spending per category split by gender

```python
# 9. Create a pivot table showing total spending per category split by gender.

df.pivot_table(values='amount', index='category', columns='gender', aggfunc='sum')

df
```

#### 10. Create a pivot table (name vs month) with total amount and replace missing values with 0 using fill_value

```python
# 10. Create a pivot table (name vs month) with total amount and replace missing values with 0 using fill_value.

df.pivot_table(values='amount', index='name', columns='month', aggfunc='sum', fill_value=0)

df
```
