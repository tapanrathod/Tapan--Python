#### 31. Create new column salary_category (Low/Medium/High)


```python
# 31. Create new column salary_category (Low/Medium/High)

import pandas as pd

df = pd.read_csv("..\dataset\pandas_full_practice.csv")


def categorize_salary(salary):

    if salary < 50000:
        return "Low"
    elif salary < 100000:
        return "Medium"
    else:
        return "High"


df["salary_category"] = df["salary"].apply(categorize_salary)

print(df[["salary", "salary_category"]])

df[["salary", "salary_category"]]
```

#### 32. Create experience_level column


```python
# 32. Create experience_level column

import pandas as pd

df = pd.read_csv("..\dataset\pandas_full_practice.csv")


def categorize_experience(experience):

    if experience < 2:
        return "Junior"
    elif experience < 5:
        return "Mid"
    else:
        return "Senior"


df["experience_level"] = df["experience"].apply(categorize_experience)

print(df[["experience", "experience_level"]])

df[["experience", "experience_level"]]
```
