#### 14. Create a DataFrame from user input for 5 students (Name and Marks).

```python
# 14. Create a DataFrame from user input for 5 students (Name and Marks).

import pandas as pd

names = []
marks = []

for i in range(5):
    name = input(f"Enter name for student {i+1}: ")
    mark = int(input(f"Enter marks for student {i+1}: "))

    names.append(name)
    marks.append(mark)

df_students = pd.DataFrame({
    "Name": names,
    "Marks": marks
})

print(df_students)
```

```python
# 60. Filter customers whose Purchase > 5000 and City is "Surat".

import pandas as pd

data = {
    "CustomerID": [1, 2, 3, 4, 5],
    "Name": ["Amit", "Riya", "Karan", "Neha", "Sita"],
    "City": ["Delhi", "Surat", "Ahmedabad", "Surat", "Kolkata"],
    "Purchase": [6000, 7000, 4000, 8000, 3000]
}

df_customers = pd.DataFrame(data)

filtered_customers = df_customers[(df_customers["Purchase"] > 5000) & (df_customers["City"] == "Surat")]

print(filtered_customers)
```
