```python
import pandas as pd

payments = pd.read_csv("./dataset/payments.csv")
payments.head()
```

```python
payments.info()

payments.shape
```

```python
df = payments.groupby(["city", "payment_method"])["amount"].sum().reset_index(name="transection")
df
```

```python
df.pivot_table(index="city", columns="payment_method", values="transection", aggfunc="sum")
```
