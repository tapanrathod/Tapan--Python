```python
import pandas as pd
```

```python
df = pd.read_csv("./dataset/payments.csv")
df.head() # to show entire table
```

```python
df.info() # must to know datatypes
```

```python
df2 = df[["city", "amount", "month", "payment_method", "gender"]] # new df for pivot table
df2.head()
```

#### pivot table
#### index, vales and columns are must be specified in pivot table
#### values must be numeric data
#### index and columns can be categorical data
#### aggregation function is used to aggregate the values in pivot table

```python
df2.pivot_table(index="month", values="amount", columns="city", aggfunc="sum")
```

```python
df.groupby(["city"])["amount"].sum()
```

```python
df.pivot_table(index={"month", "payment_method"}, values=["amount", "category"], columns=["city", "gender"], aggfunc="sum")
```
