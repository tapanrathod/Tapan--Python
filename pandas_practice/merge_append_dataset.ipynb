#### Merge

```python
import pandas as pd
```

```python
all_orders = pd.read_csv("./dataset/all_orders.csv")
customers = pd.read_csv("./dataset/customers.csv")
```

```python
all_orders.head() # to see data set
```

```python
customers.head() # to see data set
```

```python
merged_df = pd.merge(customers, all_orders) # merge (left side, right side)
merged_df.head() # print
```

```python
merged_df = pd.merge(all_orders, customers) # merge (left side, right side)
merged_df.head() # print
```

#### Left Merge

```python
merged_df = pd.merge(customers, all_orders, how="left")
merged_df # print
```

#### Left Anti Merge

```python
merged_df = pd.merge(customers, all_orders, how="left_anti")
merged_df # print
```

#### Right Merge

```python
merged_df = pd.merge(customers, all_orders, how="right")
merged_df # print
```

#### Right Anti Merge

```python
merged_df = pd.merge(customers, all_orders, how="right_anti")
merged_df
```

#### Cross Merge

```python
merged_df = pd.merge(customers, all_orders, how="cross")
merged_df # print
```

```python
merged_df = pd.merge(all_orders, customers, how="cross")
merged_df # print
```

#### Inner Merge

```python
merged_df = pd.merge(customers, all_orders, how="inner")
merged_df # print
```

#### Outer Merge

```python
merged_df = pd.merge(customers, all_orders, how="outer")
merged_df # print
```

#### Append

```python
orders1 = pd.read_csv("./dataset/orders.csv")
orders2 = pd.read_csv("./dataset/orders2.csv")
```

```python
orders1.head() # to see data set
```

```python
orders2.head() # to see data set
```

```python
append_df = pd.concat([orders1, orders2]) # order2 appends to order1
append_df # print
```

```python
append_df = pd.concat([orders1, orders2]).reset_index(drop=True)  # reset index to get index values from the beginning
append_df # print
```

#### New CSV File

```python
append_df.to_csv("appended_order1_and_order2.csv")
```
