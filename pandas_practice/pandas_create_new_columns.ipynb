#### Combining Departments and Cities

```python
df["city_department"] = df["city"] + "-" + df["department"] # here "_" is separator

df.head()
```

#### Separating Salary Categories

```python
def salary_category(x):

    if x >= 100000:
        return "High"
    
    elif x >= 50000:
        return "Medium"
    
    elif x >= 20000:
        return "Low"


df["salary_category"] = df["salary"].apply(salary_category)

df
```

```python
df["salary"].min() # minimum salary

df["salary"].max() # maximum salary
```

#### Salary Groupby

```python
df.groupby("salary_category")["salary_category"].count() # groupby salary category
```

```python
df.groupby(["salary_category", "city"])["salary_category"].count() # salary group by city
```
